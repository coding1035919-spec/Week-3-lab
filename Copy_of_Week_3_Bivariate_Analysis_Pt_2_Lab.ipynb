## Scatterplot

### What is scatterplot?
A scatterplot is used to get an insight on the relationship between two variables $X$ and $Y$. The $(x, y)$ coordinates are represented on a cartesian plane, and the plot itself can reveal patterns or correlations in the data, such as linear relationships or clusters of points.

### Advantages
* Easy to interpret
* It is a useful tool to identify potential univariate and bivariate outliers
* Can also be used to get an insight as to whether two variables have positive or negative correlation (and the extent to which this is the case).
* Can be used as a tool alongside the [Variance Inflation Factor](https://en.wikipedia.org/wiki/Variance_inflation_factor) to detect [multicollinearity](https://en.wikipedia.org/wiki/Multicollinearity) in [multiple linear regression models](https://en.wikipedia.org/wiki/Linear_regression)
* We can make our analysis multivariate in nature by colouring and sizing the points depending on a third and fourth variable etc.

## Disadvantages
* With large datasets, too many points can make the graph look messy
* Outliers (particularly $y$ outliers) can distort the relationship between the two variables
* Gives us an insight on correlation, but is not sufficient for determining causality

## The Task
Here we take a look at the relationship between infant mortality and life expectancy across a variety of different countries in the year 2018.

## Scatterplot: Plotly Express

With plotly express, we simply call the `scatter` top-level function and pass in the `x` and `y` columns as strings, assuming the `data_frame` argument references the correct data

```python
fig = px.scatter(data_frame=health, x="InfantMortality", y="Life_exp", color_discrete_sequence=['green'])

fig.update_layout(title='Scatterplot of Life_exp vs InfantMortality')

fig.show()
```

## Bubble Chart

### What is Bubblechart?
A bubble chart is a scatter plot in which the size of the scatters is determined by a third variable

### Advantages
* Can represent 3 variables on a two dimensional object
* Easy to see the relationship between the three variables
* Provides a more comprehensive view of the data since we have more variables we can extract insights from

### Disadvantages
* Big bubbles can smother smaller ones, hiding them from the graph
* Requires adequate data cleaning of the third variable (controlling size) so that the graph does not become distorted


### **Your Turn**

Look up the plotly docs for a Bubble chart and create a plot of life expectancy against infant mortality, denoting the size of the point through the corresponding value for GDP per capita.



```python
# Create a bubble chart using Plotly Express
# x-axis: InfantMortality
# y-axis: Life_exp
# size of bubbles: GDP/Capita
fig = px.scatter(data_frame=health,
                 x="InfantMortality",
                 y="Life_exp",
                 size="GDP/Capita",
                 hover_name="Country", # Show country name on hover
                 title="Life Expectancy vs Infant Mortality (Bubble size: GDP/Capita)")

# Display the plot
fig.show()
```

## Correlation Matrix & Heatmap

The Pearson's correlation coefficient (also known as Pearson's $r$) is a statistic that measures the degree to which two variables move together in
linear fashion. Correlation takes on a value between -1 and 1, in which $r=-1$ implies perfect negative correlation  (points move together in a perfect straight line with a negative gradient) and $r=1$ implies perfect positive correlation (points move together in a perfect straight line with a positive gradient)

```python
# We only have 2018 data hence we can't calculate correlation to do with year
# Drop the year column since we don't need it
corr = health.drop(columns='Year').corr(numeric_only=True)
corr
```

## Correlation Matrix: Plotly Express

Here we use the `px.imshow()` top level function to plot our correlation matrix. We pass in our correlation matrix `corr` as the first argument and set `text_auto=True` to show the computed values on the plot. The `color_continuous_scale` arguments provides us with a colour bar for colour coding purposes, though we will explore colour in more depth in the next chapter.

```python
# Plot our correlation heatmap
fig = px.imshow(corr, text_auto=True, color_continuous_scale='viridis_r')
fig.show()
```

## Line Charts and Multi Line Charts

### What is Line chart?
Line charts are particularly useful visualisation options for when we wish to visualise patterns and trends of a continuous variable over *time*

### Advantages
* Easy to visualise trends and patterns in data to see how a continuous variable changes over time
* Can plot many lines on a single chart
* Easy to identify areas of high fluctuation in the values of the data

### Disadvantages
* Struggles to capture the way the data is distributed
* Can become messy as we add to many variables within a multi-line chart
* The plot can look extremely distorted even if a single value is incorrect or is an outlier

### **Your Turn**
*   Select at least 5-10 countries from the health dataset and create a multiline chart to compare development indicators.
*   Hint: Normalise the numeric columns
*   Add appropriate title, axis labels and legend















```python
# Select 5-10 countries for comparison
selected_countries = [
    'Denmark', 'Mongolia', 'Switzerland', 'Argentina', 'Paraguay',
    'United States', 'Ethiopia', 'Japan', 'Norway', 'India'
]

# Filter the health data for selected countries
country_health = health[health['Country'].isin(selected_countries)].copy()

# Define numeric columns to normalize
numeric_cols = ['GDP/Capita', 'MeanSchooling', 'InfantMortality', 'Life_exp']

# Normalize the numeric columns using min-max scaling
for col in numeric_cols:
    min_val = country_health[col].min()
    max_val = country_health[col].max()
    country_health[f'{col}_Normalized'] = (country_health[col] - min_val) / (max_val - min_val)

# Melt the DataFrame for plotting multiple indicators
melted_health = country_health.melt(
    id_vars=['Country', 'Year'],
    value_vars=[f'{col}_Normalized' for col in numeric_cols],
    var_name='Indicator',
    value_name='Normalized Value'
)

# Create the multiline chart
fig = px.line(
    melted_health,
    x='Year',
    y='Normalized Value',
    color='Country',
    line_dash='Indicator',
    title='Normalized Development Indicators Across Selected Countries',
    labels={
        'Year': 'Year',
        'Normalized Value': 'Normalized Value (Min-Max Scaled)',
        'Country': 'Country',
        'Indicator': 'Indicator'
    }
)

fig.update_layout(
    hovermode='x unified',
    xaxis_title="Year (2018 Only)",
    yaxis_title="Normalized Value",
    legend_title="Country and Indicator"
)

fig.show()
```

### Answers to the Questions

*   **Why is normalisation important before plotting these indicators together?**
    Normalization (specifically min-max scaling, as used here) is crucial because the indicators (`GDP/Capita`, `MeanSchooling`, `InfantMortality`, `Life_exp`) have vastly different scales. For example, `GDP/Capita` can be in the tens of thousands, while `MeanSchooling` is typically a single or double-digit number. If plotted directly without normalization, the indicator with the largest scale (e.g., GDP/Capita) would dominate the chart, making it impossible to visually compare the trends and patterns of the other indicators. Normalization scales all values to a common range (e.g., 0 to 1), allowing for a fair and meaningful visual comparison of their relative changes and relationships.

*   **What patterns do you observe between GDP, schooling, and life expectancy?**
    From the normalized multiline chart, we can generally observe a positive correlation between `GDP/Capita`, `MeanSchooling`, and `Life_exp`. Countries with higher normalized GDP per capita tend to also have higher normalized mean schooling years and higher normalized life expectancy. Conversely, countries with lower GDP per capita generally show lower mean schooling and life expectancy. For `InfantMortality`, an inverse relationship is typically seen: as GDP, schooling, and life expectancy increase, infant mortality tends to decrease. This suggests that socio-economic development, educational attainment, and healthcare (reflected in infant mortality and life expectancy) are interconnected.
